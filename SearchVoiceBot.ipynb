```python
import webbrowser
import speech_recognition as sr

chrome_path = "C:/Program Files/Google/Chrome/Application/chrome.exe"

try:
    webbrowser.register('chrome', None, webbrowser.BackgroundBrowser(chrome_path))
except Exception as e:
    print(f"Error registering Chrome: {e}")

def listen_to_voice():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening for your search query...")
        audio = recognizer.listen(source)

        try:
            # Recognize speech using Google's speech recognition
            query = recognizer.recognize_google(audio)
            print(f"You said: {query}")
            return query
        except sr.UnknownValueError:
            print("Sorry, I could not understand the audio.")
        except sr.RequestError as e:
            print(f"Could not request results from Google Speech Recognition: {e}")

    return None

def search_in_chrome(query):
    if query:
        try:
            url = f"https://www.google.com/search?q={query}"
            print(f"Opening URL: {url}")
            webbrowser.get("chrome").open(url)
        except Exception as e:
            print(f"Error opening Chrome: {e}")

if __name__ == "__main__":
    # Listen for the voice query
    voice_query = listen_to_voice()
    # Search in Chrome using the voice query
    search_in_chrome(voice_query)
```
